# PowerCo Churn Analysis – Exploratory Data Analysis (EDA)

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```
# Load data files
client_df = pd.read_csv("client_data1.csv")
price_df = pd.read_csv("price_data1.csv")

# Display first few rows
client_df.head()
```

```
# Display data types
print("Client Data Types:\n", client_df.dtypes)
print("\nPrice Data Types:\n", price_df.dtypes)
```

```
# Summary statistics for client and price data
client_df.describe(include='all')
```

```
price_df.describe(include='all')
```

```
# Check for missing values
print(client_df.isnull().sum())
print(price_df.isnull().sum())
```

```
# Value counts for categorical variables
categorical_cols = client_df.select_dtypes(include='object').columns
for col in categorical_cols:
    print(f"Distribution for {col}:\n")
    print(client_df[col].value_counts(), "\n")
```

```
# Histograms for numeric columns
numeric_cols = client_df.select_dtypes(include=['int64', 'float64']).columns
client_df[numeric_cols].hist(figsize=(15, 12), bins=30)
plt.tight_layout()
plt.suptitle("Numeric Column Distributions", y=1.02)
plt.show()
```

```
# Boxplot of net margin
plt.figure(figsize=(8, 4))
sns.boxplot(x=client_df["net_margin"])
plt.title("Boxplot of Net Margin")
plt.show()
```

```
# Churn distribution
sns.countplot(x="churn", data=client_df)
plt.title("Churn Distribution")
plt.show()
```
